```
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```
# Data set creation
np.random.seed(42)
a = np.random.randint(1,100,size=1000)
b = np.random.randint(1,100,size=1000)
c = a + b

features = np.column_stack((a,c)) # for input
labels = b # for output
```

```
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42)
```

```
model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(2,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
```

```
history = model.fit(x_train, y_train, epochs=50, batch_size=32, validation_split=0.2, verbose=1)
```

```
# Custom input values for a and c
custom_input = np.array([[1, 2], [3, 5]])  # Each row is a pair of (a, c)

# Make predictions
predictions = model.predict(custom_input)

# Print predictions
for i, pred in enumerate(predictions):
    print(f"Custom Input (a, c): {custom_input[i]}, Predicted b: {pred[0]:.2f}")
```
